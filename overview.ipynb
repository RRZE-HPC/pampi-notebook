## Working with the Material

The material can be accessed in different ways, depending on what your preferences are.
* Follow the material in the Jupyter notebooks and run the examples directly in the notebook.
  The C code cells are compiled with `mpicc` and started with `mpirun` by the `%%mpi` cell magic, which is explained at the beginning of each notebook.
* You can also work on the terminal
    * `file` > `new` > `terminal`
    * The source code files for all examples discussed are located at `material/code`
